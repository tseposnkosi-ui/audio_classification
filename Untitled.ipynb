```python
import os
import torch
import torchaudio
from torch import nn
from torchaudio.transforms import Resample, Spectrogram, MelSpectrogram
from torch.utils.data import Dataset, DataLoader, random_split
```

```python
# Pipeline for transforming audio data into spectrograms and statistical data
class MyPipeline(nn.Module):
    def __init__(
        self,
        input_freq=48000,
        resample_freq=8000,
        window_len=256,
        hop_len=100,
        number_mel=64
    ):
        super().__init__()
        self.resample = Resample(orig_freq=input_freq, new_freq=resample_freq)
        self.spec     = Spectrogram(win_length=window_len, hop_length=hop_len, power=2)
        self.mel_spec = MelSpectrogram(sample_rate=resample_freq, win_length=window_len, hop_length=hop_len, n_mels=number_mel)

    def forward(self, waveform: torch.Tensor):
        
        resampled       = self.resample(waveform)
        # spectrograms
        spectrogram     = self.spec(resampled).squeeze(0)
        mel_spectrogram = self.mel_spec(resampled).squeeze(0)
        
        # statistical data
        averagePower   = torch.mean(torch.square(resampled))
        mean           = torch.mean(resampled)
        vari           = torch.var(resampled)
        rootMeanSquare = torch.sqrt(torch.mean(torch.square(resampled)))
        signs          = torch.sign(resampled)
        crossings      = (signs[:, 1:] != signs[:, :-1])
        zeroCrossings  = torch.sum(crossings).float()
        peakToPeak     = torch.max(resampled) - torch.min(resampled)
        # putting the statistical data together
        statFeatures = torch.stack([averagePower, mean, vari, rootMeanSquare, zeroCrossings, peakToPeak])

        return spectrogram, mel_spectrogram, statFeatures
```

```python
# The data loader function
class AudioDataset(Dataset):
    
    def __init__(self, root_dir, pipeline, allowed_sources=None):
        self.pipeline = pipeline
        self.data = []
        for source in sorted(os.listdir(root_dir)):
            if allowed_sources is not None and source not in allowed_sources:
                continue
            source_path = os.path.join(root_dir, source)
            if not os.path.isdir(source_path):
                continue
            for file in os.listdir(source_path):
                if not file.endswith(".wav"):
                    continue
                try:
                    label = int(file.split("_")[0])
                    self.data.append((os.path.join(source_path, file), label))
                except (ValueError, IndexError):
                    print(f"Skipping malformed filename: {file}")
                    
    def __len__(self):
        return len(self.data)
        
    def __getitem__(self, idx):
        path, label = self.data[idx]
        waveform, sample_rate = torchaudio.load(path)
        max_len = 32000
        if waveform.shape[1] < max_len:
            waveform = torch.nn.functional.pad(
                waveform,
                (0, max_len - waveform.shape[1])
            )
        else:
            waveform = waveform[:, :max_len]
            
        spec, mel_spec, statFeatures = self.pipeline(waveform)
        
        return spec, mel_spec, statFeatures, label
```

```python
# Model 
class CNNNet(nn.Module):
    def __init__(self, num_classes=10):
        super(CNNNet, self).__init__()

        self.features_1 = nn.Sequential(
            nn.Conv2d(1,   32,  kernel_size=(3,3)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2,2)),
            nn.Conv2d(32,  64,  kernel_size=(3,3)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2,2)),
            nn.Conv2d(64,  128, kernel_size=(3,3)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2,2)),
        )
        self.features_2 = nn.Sequential(
            nn.Conv2d(1,   32,  kernel_size=(3,3)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2,2)),
            nn.Conv2d(32,  64,  kernel_size=(3,3)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2,2)),
            nn.Conv2d(64,  128, kernel_size=(3,3)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2,2)),
        )

        self.mlp_1 = nn.Sequential(
            nn.Linear(6, 64),  nn.ReLU(),
            nn.Linear(64, 64), nn.ReLU(),
            nn.Linear(64, 32), nn.ReLU(),
        )

        self.mlp_2_1 = nn.Sequential(
            nn.Flatten(),
            nn.LazyLinear(128),
            nn.ReLU(),
        )
        self.mlp_2_2 = nn.Sequential(
            nn.Flatten(),
            nn.LazyLinear(128),
            nn.ReLU(),
        )

        self.mlp_3 = nn.Sequential(
            nn.LazyLinear(128),
            nn.ReLU(),
            nn.Linear(128, num_classes),
            nn.Softmax(dim=1),
        )

    def forward(self, x1, x2, x3):
        x1 = x1.unsqueeze(1)   # [batch, 1, freq, time]
        x2 = x2.unsqueeze(1)   # [batch, 1, freq, time]

        x1_conv = self.features_1(x1)
        x2_conv = self.features_2(x2)
        x3_mlp  = self.mlp_1(x3)

        x1_conv_mlp = self.mlp_2_1(x1_conv)
        x2_conv_mlp = self.mlp_2_2(x2_conv)

        combinedLayer  = torch.cat([x1_conv_mlp, x2_conv_mlp, x3_mlp], dim=1)
        classification = self.mlp_3(combinedLayer)
        return classification
```

```python
# Setting up things
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
pipeline = MyPipeline()
dataset  = AudioDataset(root_dir="data", pipeline=pipeline)
train_size = int(0.8 * len(dataset))
val_size   = len(dataset) - train_size
train_dataset, val_dataset = random_split(dataset, [train_size, val_size])
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True,  num_workers=2)
val_loader   = DataLoader(val_dataset,   batch_size=32, shuffle=False, num_workers=2)
model     = CNNNet(num_classes=10).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
# training loop
def train(model, train_loader, val_loader,
          criterion, optimizer, epochs=20):
    
    for epoch in range(epochs):
        
        model.train()
        train_loss = 0.0
        train_correct = 0
        train_total = 0
        
        for specs, mel_specs, stat_features, labels in train_loader:
            
            specs = specs.to(device)
            mel_specs = mel_specs.to(device)
            stat_features = stat_features.to(device)
            labels = labels.to(device)
            optimizer.zero_grad()
            outputs = model(specs, mel_specs, stat_features)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
            preds = outputs.argmax(dim=1)
            train_correct += (preds == labels).sum().item()
            train_total += labels.size(0)
            
        model.eval()
        
        val_loss = 0.0
        val_correct = 0
        val_total = 0
        
        with torch.no_grad():
            
            for specs, mel_specs, stat_features, labels in val_loader:
                
                specs = specs.to(device)
                mel_specs = mel_specs.to(device)
                stat_features = stat_features.to(device)
                labels = labels.to(device)
                
                outputs = model(specs, mel_specs, stat_features)
                
                loss = criterion(outputs, labels)
                
                val_loss += loss.item()
                
                preds = outputs.argmax(dim=1)
                
                val_correct += (preds == labels).sum().item()
                val_total += labels.size(0)
                
        print(
            f"Epoch {epoch+1}/{epochs} | "
            f"Train Loss: {train_loss/len(train_loader):.4f} | "
            f"Train Acc: {train_correct/train_total:.4f} | "
            f"Val Loss: {val_loss/len(val_loader):.4f} | "
            f"Val Acc: {val_correct/val_total:.4f}"
            )
```

```python
train(model, train_loader, val_loader, criterion, optimizer, epochs=20)
```
